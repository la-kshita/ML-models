```python
import pandas as pd
```

```python
df = pd.DataFrame([['green', 'M', 10.1, 'class2'],
                   ['red', 'L', 13.5, 'class1'],
                   ['blue', 'XL', 15.3, 'class2']])
df.columns = ['color', 'size', 'price', 'classlabel']
```

Nominal feature : colour
Ordinal feature : Size
Numerical feature : Price
Class labels

```python
df
```

```python
size_mapping = {'XL': 3,
                'L': 2,
                'M': 1}
```

```python
df['size'] = df['size'].map(size_mapping)
```

```python
df
```

```python
# To convert the size back into strings we can define a reverse dictionary comprehension
inv_size_mapping = {v: k for k, v in size_mapping.items()}
```

```python
df['size'] = df['size'].map(inv_size_mapping)
```

```python
df
```

```python
# Since a lot of ML algorithms require class labels to be numbers , hence we convert them. Since they are not ordinal thus can be converted using enumerate
```

```python
class_mapping = {label: idx for idx, label in enumerate(np.unique(df['classlabel']))}
```

```python
class_mapping
```

```python
df['classlabel'] = df['classlabel'].map(class_mapping)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
class_le = LabelEncoder()
y = class_le.fit_transform(df['classlabel'].values)
y
```

```python
from sklearn.preprocessing import OneHotEncoder
X = df[['color','size','price']].values
color_ohe = OneHotEncoder()
color_ohe.fit_transform(X[:,0].reshape(-1,1)).toarray()
```

```python
from sklearn.compose import ColumnTransformer
```

```python
X = df[['color','size', 'price']].values
c_transf = ColumnTransformer([('onehot', OneHotEncoder(), [0]),('nothing','passthrough',[1,2])])
c_transf.fit_transform(X).astype(float)
```
