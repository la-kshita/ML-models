```python
X= df.drop(columns = ['Purchased'])
y = df['Purchased']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state = 0)

X_train.shape, y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler

#StandardScaler is not a function instead it is a class
scaler = StandardScaler()

#Fit the scaler into the values - Understands the dataset and learns the values like the mean and standard deviation
scaler.fit(X_train)

#Transforming both the training set and the testing set
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```
